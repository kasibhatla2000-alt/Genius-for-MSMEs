# Genius v5.1: separate Genius and Grit scores
**BMF5342 · an intrapreneur proposal for Zoho**

**Genius** measures current financial credit standing using learned financial coefficients. **Grit** measures resilience using demonstrated recovery and capacity. They stay separate: overlapping financial inputs and an unvalidated blend weight would make a combined score hard to defend.

**Borrow is request-based:** enter amount, product and duration for an illustrative quote, full fees, cash received and repayment schedule.

**Recovery policy:** completed, recent recovery evidence can earn a small margin concession under an explicitly assumed demo policy. It changes neither Genius nor calibrated PD and adds no size bonus. Cash buffers and debt cover are not rewarded twice. With no completed evidence, there is no concession.

**Synthetic resilience example:** Borrow → Demo: how resilience can affect credit pricing. Two constructed histories face equal shocks but respond differently; their Grit scores are calculated by the existing detector. Same current finances, Genius 65 and ₹5 lakh / 60 days: Grit about 74 vs 54, rates about 14.82% vs 14.95%. These are scenarios, not validation on real loans. Historical default labels have not been rewritten.

**Run:** Runtime → Run all. The final cell prints a temporary public link while the runtime remains running.

## 2 · Build the synthetic demonstration and train the financial score

- 5,000 firms in 10 sectors; the simulator includes recovery speed, cost flexibility, cash buffers and shocks.
- Monotone financial risk coefficients replace fixed area weights in Genius. Data quality gates publication.
- Genius is financial-only; Grit remains separate. The combined candidate is reported for research, not deployed.
- Training feature months end at 47, with labels known by 59; validation ends at 83, labels known by 95. Final test features are 96–107.
- A separate recovery concession is a lender-policy assumption, not a learned default-risk reduction. It uses only completed recovery features, not Grit capacity; no amount bonus or financial-gate override applies.
- Extra constructed resilience histories are supplied as operating signals, not accounting ledger exports.

Synthetic records demonstrate implementation; real repayment data and lender validation are still required.

## 4 · The three sample businesses, end to end
The same code the app uses: read the books, check them, score, compare, Grit, Second Opinion, limits and rates.

```python
#@title Assess Sharma Traders, Kaveri Foods and Swift Freight { display-mode: "form" }
from genius.session import Session
rows, SESS = [], {}
for label, (fn, cc, mse) in app.SAMPLES.items():
    t = time.time()
    s = Session(); r = s.read(f"samples/{fn}"); a = s.run(DB, app.new_policy(), r["sector"], cc, mse)
    SESS[label] = s
    g = a.grit["score"][0, -1]
    rows.append(dict(Business=label.split(" (")[0], Sector=a.sector, Size=a.band, Score=None if a.score is None else round(a.score, 1), Tier=a.tier,
                     Peers=f"rank {a.peers['rank']} of {a.peers['n']}", Grit=None if np.isnan(g) else round(float(g)),
                     **{"Second Opinion (lender only)": a.so.get("confidence"),
                        "Short-term limit": app.ui.money(a.limits.get("short", {}).get("limit", 0)) if a.limits and a.limits["short"]["available"] else "not offered",
                        "Rate": f"{a.pricing['short']['rate']:.2f}%" if a.tier and a.tier != "Risky" else "-", "Seconds": round(time.time() - t)}))
display(pd.DataFrame(rows))
```

```python
#@title Ask Genius without any AI model: typed questions matched to a bank of answerable questions { display-mode: "form" }
s = SESS[list(SESS)[0]]
for q in ["why is my score low", "how much can i borrow short term", "who owes me the most", "what if gupta hardware pays",
          "what is my dso", "explain grit", "is the peer data real", "what's the weather"]:
    ans = s.genie.ask(q)
    print(f"Q: {q}\n   [{ans.status}] {ans.understood or ''}\n   {ans.text[:300]}\n")
print("Menu questions:", len(s.genie.bank))
```

## Source and limitations

- `credit_model.py`: financial-only Genius, with exploratory combined-model results kept separate.
- `grit.py`: original demonstrated recovery and capacity score; weights remain assumptions.
- `resilience.py`: completed-evidence gating and the assumed recovery concession.
- `resilience_demo.py`: constructed fast/slow recovery histories and a fixed-finances pricing comparison.
- `samples/synthetic_resilience_histories.csv`: reproducible operating signals, not a ledger upload.
- `lending.py` / `policy.py`: requested amount, stressed affordability, full costs and editable lender policy.
- `app.py` / `ui.py`: both scores, request-based Borrow and clearly labelled scenario comparison.

No live bank/Zoho/GST/bureau feeds, lender approvals or disbursements. The concession is an unvalidated policy assumption; synthetic model metrics do not establish its real-world value. Effective annual cost is an illustration, not a regulatory KFS APR.
